### Objectives
- Load the dataset
- Handle missing values
- Remove duplicates
- Create missing value indicator features
- Clean and normalize text
- Create text statistics
- Save the processed dataset

```python
import pandas as pd
import numpy as np

import re
import string
import nltk

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

from IPython.display import display

print("Libraries imported successfully!")
```

```python
nltk.download("punkt")
nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("omw-1.4")

print("NLTK resources are ready!")
```

```python
df = pd.read_csv("../data/raw/fake_job_postings.csv")

print("Dataset loaded successfully!")

print("Rows :", df.shape[0])
print("Columns :", df.shape[1])

display(df.head())
```

```python
print("="*60)
print("Dataset Information")
print("="*60)

df.info()

print("\nMissing Values")

display(
    df.isnull()
      .sum()
      .sort_values(ascending=False)
)
```

```python
df = df.dropna(subset=["description"])

print("Rows after removing missing description:", len(df))
```

```python
indicator_columns = [
    "salary_range",
    "department",
    "company_profile",
    "requirements",
    "benefits"
]

for col in indicator_columns:
    df[col + "_missing"] = df[col].isna().astype(int)

df[[c + "_missing" for c in indicator_columns]].head()
```

```python
fill_values = {
    "salary_range": "Not Specified",
    "department": "Not Specified",
    "company_profile": "",
    "requirements": "",
    "benefits": "",
    "employment_type": "Unknown",
    "required_experience": "Unknown",
    "required_education": "Unknown",
    "industry": "Unknown",
    "function": "Unknown",
    "location": "Unknown"
}

df.fillna(fill_values, inplace=True)

print("Missing values handled successfully.")
```

```python
df.isnull().sum()
```

```python
df.drop(columns=["job_id"], inplace=True)

print("job_id removed successfully.")
```

```python
text_columns = [
    "title",
    "company_profile",
    "description",
    "requirements",
    "benefits"
]

df["combined_text"] = df[text_columns].agg(" ".join, axis=1)

df["combined_text"].head()
```

```python
import re
import string

stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()

def clean_text(text):

    text = text.lower()

    # Remove HTML tags
    text = re.sub(r"<.*?>", " ", text)

    # Remove URLs
    text = re.sub(r"http\S+|www\S+", " ", text)

    # Remove email addresses
    text = re.sub(r"\S+@\S+", " ", text)

    # Remove numbers
    text = re.sub(r"\d+", " ", text)

    # Remove punctuation
    text = text.translate(str.maketrans("", "", string.punctuation))

    # Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    # Tokenize
    words = text.split()

    # Remove stopwords and lemmatize
    words = [
        lemmatizer.lemmatize(word)
        for word in words
        if word not in stop_words
    ]

    return " ".join(words)

df["clean_text"] = df["combined_text"].apply(clean_text)

df[["combined_text", "clean_text"]].head()
```

```python
# ==========================================================
# Create Text Statistics
# ==========================================================

df["text_length"] = df["clean_text"].str.len()

df["word_count"] = df["clean_text"].apply(lambda x: len(x.split()))

df["avg_word_length"] = (
    df["clean_text"]
      .apply(
          lambda x: np.mean([len(word) for word in x.split()])
          if len(x.split()) > 0 else 0
      )
)

print("Text statistics created successfully.")

df[
    [
        "text_length",
        "word_count",
        "avg_word_length"
    ]
].head()
```

```python
# ==========================================================
# Select Final Features
# ==========================================================

processed_df = df[
    [
        "clean_text",

        "text_length",
        "word_count",
        "avg_word_length",

        "telecommuting",
        "has_company_logo",
        "has_questions",

        "employment_type",
        "required_experience",
        "required_education",
        "industry",
        "function",

        "salary_range_missing",
        "department_missing",
        "company_profile_missing",
        "requirements_missing",
        "benefits_missing",

        "fraudulent"
    ]
]

print("Final dataset created successfully.")

processed_df.head()
```

```python
# ==========================================================
# Dataset Information
# ==========================================================

print("=" * 60)
print("Processed Dataset Information")
print("=" * 60)

processed_df.info()

print("\nShape of Dataset")

print(processed_df.shape)
```

```python
# ==========================================================
# Save Processed Dataset
# ==========================================================

import os

os.makedirs("../data/processed", exist_ok=True)

processed_df.to_csv(
    "../data/processed/processed_jobs.csv",
    index=False
)

print("=" * 60)
print("Processed dataset saved successfully!")
print("=" * 60)
print("Location: ../data/processed/processed_jobs.csv")
```

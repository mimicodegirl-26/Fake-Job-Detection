```python
# ============================================================
# 12.5 EXPLAINABLE AI (SHAP)
# ============================================================

def get_feature_names(tfidf_vectorizer, onehot_encoder, numerical_cols):
    """Reconstruct feature names from the preprocessing pipelines."""
    text_features = tfidf_vectorizer.get_feature_names_out()
    cat_features = onehot_encoder.get_feature_names_out()
    num_features = np.array(numerical_cols)
    
    return np.concatenate([text_features, cat_features, num_features])

def generate_shap_plot(model, features, feature_names):
    """Generate a SHAP waterfall plot for the single prediction."""
    dense_features = features.toarray()
    
    try:
        explainer = shap.TreeExplainer(model)
        shap_values = explainer(dense_features)
    except Exception:
        explainer = shap.Explainer(model, dense_features)
        shap_values = explainer(dense_features)

    if len(shap_values.shape) == 3:
        shap_values_target = shap_values[0, :, 1]
    else:
        shap_values_target = shap_values[0]

    fig, ax = plt.subplots(figsize=(10, 6))
    fig.patch.set_facecolor('#091522') 
    ax.set_facecolor('#091522')
    ax.tick_params(colors='white')
    ax.xaxis.label.set_color('white')
    ax.yaxis.label.set_color('white')

    shap.plots.waterfall(
        shap_values_target,
        max_display=10, 
        show=False
    )
    
    for text in plt.gca().texts:
        text.set_color('white')

    plt.tight_layout()
    return fig
```
